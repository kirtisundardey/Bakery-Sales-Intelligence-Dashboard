## Data Cleaning

```python
# Remove duplicates and create a fresh copy
df = df.drop_duplicates().copy()
```

```python
# Convert date_time column
df['date_time'] = pd.to_datetime(df['date_time'])
```

```python
# Create New Time Columns
df['Date'] = df['date_time'].dt.date
df['Month'] = df['date_time'].dt.month
df['Hour'] = df['date_time'].dt.hour
df['Day_Name'] = df['date_time'].dt.day_name()
```

```python
# Standardize Text Columns
df['Item'] = df['Item'].str.strip().str.title()
df['period_day'] = df['period_day'].str.strip().str.lower()
df['weekday_weekend'] = df['weekday_weekend'].str.strip().str.lower()
```

```python
# Save cleaned dataset
df.to_csv('data/cleaned/cleaned_bakery_sales.csv', index=False)
```
